```python
# 1. Import libraries
import numpy as np
import scipy.stats as stats

# 2. Define parameters
p = 0.3
n = 10000

# 3. Generate Bernoulli random variables
rng = np.random.default_rng(42)
data = rng.binomial(1, p, size=n)

# 4. Calculate theoretical probabilities
p_fraud = stats.bernoulli.pmf(1, p)
p_genuine = stats.bernoulli.pmf(0, p)

print("Theoretical Probability of Fraud:", p_fraud)
print("Theoretical Probability of Genuine:", p_genuine)

# 5. Calculate theoretical mean and variance
mean = stats.bernoulli.mean(p)
variance = stats.bernoulli.var(p)

print("\nTheoretical Mean:", mean)
print("Theoretical Variance:", variance)

# 6. Count fraudulent and genuine transactions
fraud_count = np.sum(data == 1)
genuine_count = np.sum(data == 0)

print("\nNumber of Fraudulent Transactions:", fraud_count)
print("Number of Genuine Transactions:", genuine_count)

# 7. Calculate simulated probabilities
simulated_fraud = np.mean(data == 1)
simulated_genuine = np.mean(data == 0)

print("\nSimulated Probability of Fraud:", simulated_fraud)
print("Simulated Probability of Genuine:", simulated_genuine)

# 8. Compare theoretical and simulated probabilities
print("\nComparison:")
print("Theoretical Fraud Probability:", p)
print("Simulated Fraud Probability:", round(simulated_fraud, 4))
print("Theoretical Genuine Probability:", 1 - p)
print("Simulated Genuine Probability:", round(simulated_genuine, 4))
```

```python
# 1. Import libraries
import numpy as np
import scipy.stats as stats

# 2. Define parameters
p = 0.3          # Probability of success
n = 10000        # Number of experiments

# 3. Generate Bernoulli random variables
rng = np.random.default_rng(42)
data = rng.binomial(n=1, p=p, size=n)

# 4. Calculate theoretical probability
p_failure = stats.bernoulli.pmf(0, p)
p_success = stats.bernoulli.pmf(1, p)

print("Theoretical Probability of Failure:", p_failure)
print("Theoretical Probability of Success:", p_success)

# 5. Calculate theoretical mean and variance
mean = stats.bernoulli.mean(p)
variance = stats.bernoulli.var(p)

print("\nTheoretical Mean:", mean)
print("Theoretical Variance:", variance)

# 6. Calculate simulated probabilities
simulated_success = np.mean(data == 1)
simulated_failure = np.mean(data == 0)

print("\nSimulated Probability of Success:", simulated_success)
print("Simulated Probability of Failure:", simulated_failure)

# 7. Count successes and failures
success_count = np.sum(data == 1)
failure_count = np.sum(data == 0)

print("\nNumber of Successes:", success_count)
print("Number of Failures:", failure_count)
```
