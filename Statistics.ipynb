```python
marks = [45, 50, 52, 55, 55, 58, 60, 60, 60, 62,
         65, 65, 68, 70, 70, 72, 75, 78, 80, 90]
```

**Measure of Central Tendency**

Calculates the mean, median, and mode.

```python
import numpy as np
from scipy import stats

mean = np.mean(marks)
median = np.median(marks)
mode = stats.mode(marks, keepdims=False).mode

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

**Measure of Dispersion**

Calculates the range, quartiles, interquartile range, variance, and standard deviation.

```python
import numpy as np

data_range = np.max(marks) - np.min(marks)
q1 = np.percentile(marks, 25)
q3 = np.percentile(marks, 75)
iqr = q3 - q1

population_variance = np.var(marks)
population_std = np.std(marks)

sample_variance = np.var(marks, ddof=1)
sample_std = np.std(marks, ddof=1)

print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Range:", data_range)
print("First Quartile (Q1):", q1)
print("Third Quartile (Q3):", q3)
print("Interquartile Range (IQR):", iqr)
print("Population Variance:", population_variance)
print("Population Standard Deviation:", population_std)
print("Sample Variance:", sample_variance)
print("Sample Standard Deviation:", sample_std)
```

**Graphical Representation**

This code creates three separate graphs: a histogram, a bar chart,box plot,Scalar plot and aKDE plot (Kernel Density Estimation).

```python
import matplotlib.pyplot as plt

plt.hist(marks, bins=6, color='skyblue', edgecolor='black')

plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
from collections import Counter

# Calculate frequency of each mark
frequency = Counter(marks)

x = sorted(frequency.keys())
y = [frequency[mark] for mark in x]

# Create bar chart
plt.figure(figsize=(12, 6))

bars = plt.bar(
    x, y,
    color='cornflowerblue',
    edgecolor='black',
    width=2.5
)

# Add frequency labels above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.05,
        str(int(height)),
        ha='center',
        va='bottom',
        fontsize=10
    )

# Add title and axis labels
plt.title('Bar Chart of Student Marks', fontsize=16, fontweight='bold')
plt.xlabel('Student Marks', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

# Format axes
plt.xticks(x, rotation=45)
plt.yticks(range(0, max(y) + 2))
plt.ylim(0, max(y) + 0.8)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt



plt.boxplot(marks, patch_artist=True,
            boxprops=dict(facecolor='lightgreen'),
            medianprops=dict(color='red', linewidth=2))

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt

students = range(1, len(marks) + 1)

plt.figure(figsize=(10, 6))

plt.scatter(students, marks, color='blue',
            s=80, edgecolor='black', alpha=0.8)

plt.title("Scatter Plot of Student Marks",
          fontsize=15, fontweight='bold')
plt.xlabel("Student Number")
plt.ylabel("Marks")
plt.xticks(students)
plt.yticks(range(40, 101, 10))
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.kdeplot(
    x=marks,
    fill=True,
    color='purple',
    linewidth=2,
    bw_adjust=1
)

plt.title("KDE Plot of Student Marks",
          fontsize=15, fontweight='bold')
plt.xlabel("Marks")
plt.ylabel("Density")
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
from scipy import stats

skewness = stats.skew(marks, bias=False)
excess_kurtosis = stats.kurtosis(marks, bias=False)

print("Skewness:", skewness)
print("Excess Kurtosis:", excess_kurtosis)

if skewness > 0:
    print("Distribution is positively skewed.")
elif skewness < 0:
    print("Distribution is negatively skewed.")
else:
    print("Distribution is approximately symmetric.")

if excess_kurtosis > 0:
    print("Distribution is leptokurtic relative to normal.")
elif excess_kurtosis < 0:
    print("Distribution is platykurtic relative to normal.")
else:
    print("Distribution has normal-like excess kurtosis.")
```
