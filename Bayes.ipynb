```python

# Import libraries
import numpy as np
import pandas as pd

# Given parameters
rng = np.random.default_rng(42)
N = 10_000

p_fraud = 0.08
p_flag_fraud = 0.85
p_flag_genuine = 0.10

# (a) Simulate true fraud status and system flag
fraud = rng.random(N) < p_fraud

flag = np.where(
    fraud,
    rng.random(N) < p_flag_fraud,
    rng.random(N) < p_flag_genuine
)

# (b) Construct the contingency table
table = pd.crosstab(
    np.where(fraud, "Fraudulent", "Genuine"),
    np.where(flag, "Flagged", "Not flagged"),
    rownames=["Actual status"],
    colnames=["System decision"],
    margins=True
)

print("Contingency Table:")
print(table)

# (c) Estimate probabilities from simulation
sim_p_flag = flag.mean()
sim_p_fraud_flag = fraud[flag].mean()

# (d) Calculate theoretical probabilities
p_genuine = 1 - p_fraud

# Law of total probability
th_p_flag = (
    p_flag_fraud * p_fraud
    + p_flag_genuine * p_genuine
)

# Bayes' theorem
th_p_fraud_flag = (
    p_flag_fraud * p_fraud / th_p_flag
)

# Compare simulated and theoretical probabilities
results = pd.DataFrame(
    {
        "Simulated": [
            sim_p_flag,
            sim_p_fraud_flag
        ],
        "Theoretical": [
            th_p_flag,
            th_p_fraud_flag
        ]
    },
    index=[
        "P(Flagged)",
        "P(Fraud | Flagged)"
    ]
)

print("\nProbability Comparison:")
print(results.round(4))
```
